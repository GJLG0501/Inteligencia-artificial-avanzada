```python
import polars as pl
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import plotly.express as px
import altair as alt
```

```python
#Configuración de estilo para gráficos
sns.set_style("whitegrid")
plt.rcParams['figure.figsize'] = (12, 8)
# Desabilitar la advertencia de estilo de Altair
alt.data_transformers.enable('default', max_rows=None)
print("Librerías importadas y configuradas correctamente.")
```

```python
# carga y limpieza de datos
url = "https://data.insideairbnb.com/mexico/df/mexico-city/2026-06-15/data/listings.csv.gz"
print("Intentando cargar datos desde la URL:", url)
df = pl.DataFrame()
try:
    import urllib.request
    import gzip
    import io

    with urllib.request.urlopen(url, timeout=30) as response:
        data = response.read()

    with gzip.open(io.BytesIO(data), mode="rt", encoding="utf-8") as f:
        df = pl.read_csv(f)

    print(f"Dataset cargado exitosamente. Filas: {df.shape[0]}, columnas: {df.shape[1]}")
except Exception as e:
    print(f"Error al cargar el dataset: {e}")
    print("Verifica tu conexión a internet y que la URL sea accesible. Si no hay red, descarga el archivo localmente y carga desde disco.")

# limpieza inicial
if df is not None and not df.is_empty():
    df = df.with_columns(
        pl.col("price")
        .str.replace_all(r"\$", "")
        .str.replace_all(r",", "")
        .cast(pl.Float64, strict=False)
        .alias("price")
    ).drop_nulls(
        subset=["price"]
    ).filter(
        pl.col("price") > 0
    )
    print("Limpieza inicial realizada")
    display(df.head())
else:
    print("No se realizó la limpieza porque no se pudo cargar el dataset.")
```

```python
#Analisis basico
print("1. Precio promedio por noche?")
avg_price = df["price"].mean()
display(avg_price)
print("2. Cuales son los tipos de alojamieonto")
room_types = df.group_by('room_type').agg(pl.len().alias('count')).sort('count', descending=True)
print(room_types)
print("3. Cuales son las 10 alcadias con mas alojamientos?")
top_10_alcadias = df.group_by('neighbourhood_cleansed').agg(pl.len().alias('count')).sort('count', descending=True).head(10)
print(top_10_alcadias)
```
