```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import requests
import zipfile
import io
```

```python
url = "https://ecobici.cdmx.gob.mx/wp-content/uploads/2026/08/public_data_web_2026-07.csv"
csv_file_name="2026-07.csv"
print("\n Fase 1: Extracción de datos")
try:
    response = requests.get(url)
    response.raise_for_status()  # Check for HTTP errors
    print("Descarga exitosa del archivo CSV.")
except requests.exceptions.Timeout:
    print("Error: La solicitud ha excedido el tiempo de espera.")
except requests.exceptions.RequestException as e:
    print(f"Error al descargar el archivo CSV: {e}")
    df_raw=pd.DataFrame()  # Create an empty DataFrame in case of error
```

```python
with open(csv_file_name, 'wb') as f:
    f.write(response.content)
print(f"Archivo {csv_file_name} descargado y guardado correctamente.")

print(f"Lyendo el archivo CSV {csv_file_name}...")
df_raw = pd.read_csv(csv_file_name)
print("Archivo CSV leído correctamente.")
print(f"Se cargaron{df_raw.shape[0]} filas y {df_raw.shape[1]} columnas.")
display(df_raw.head(80))
```
