```python
import polars as pl
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import plotly.express as px
import altair as alt
```

```python
#Configuración de estilo para gráficos
sns.set_style("whitegrid")
plt.rcParams['figure.figsize'] = (12, 8)
# Desabilitar la advertencia de estilo de Altair
alt.data_transformers.enable('default', max_rows=None)
print("Librerías importadas y configuradas correctamente.")
```

```python
url= "https://data.insideairnbnb.com/mexico/df/mexico-city/2026-06-15/data/listings.csv.gz"
print("Cargando datos desde la URL:", url)
df = pl.DataFrame()  # inicializar por si falla la carga
try:
    df = pl.read_csv(url)
    print("Datos cargados correctamente. Número de filas:", df.shape[0], "Número de columnas:", df.shape[1])
except Exception as e:
    print("Error al cargar los datos:", e)

#Limpieza de datos
if not df.is_empty():
    print("Realizando limpieza de datos...")
    # Eliminar filas con valores nulos en columnas críticas
    df = df.with_columns([
        pl.col("price")
        .str.replace_all(r"[$,]", "")
        .str.replace_all(",","")
        .cast(pl.Float32)
        .alias("price")
    ]).drop_nulls(
        subset=["price"]
    ).filter(
        pl.col("price") > 0
    )
    
    print("Limpieza de datos completada. Número de filas restantes:", df.shape[0])
```
